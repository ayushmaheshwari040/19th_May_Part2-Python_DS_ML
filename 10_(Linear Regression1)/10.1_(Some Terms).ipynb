# Challanges In ML

1. Insufficient Data
2. Non Representable Data
   - Data should be representable to the target
   - Ex:- Area, Location, Rooms, Age, etc affects the Price of House.
3. Biasing
4. Data Leakage
   - duplicacy/duplicate of data

# Underfitting & Overfitting

- Underfitting-> When the training score is less than the testing score.
- Overfitting-> when the training score is more than the testing scre.

# Sampling Techninques

1. Sequential Sampling
   - Starting 80% in training & 20% in testing.
   - This is default sampling.
   - Here we found the 100% chance of biasing.
2. Random Sampling
   - Randomly select 80% data.
3. Statified Sampling
   - Divide the data on the basis of coulums by checking the proportion of data so that equal ratio of data in training and testing data.
