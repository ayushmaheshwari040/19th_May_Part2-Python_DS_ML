```python
height=np.random.normal(150,20,100)             #(centre of data, deviation, no. of data )
height
```

```python
weight=0.6*height +10
weight
```

```python
np.corrcoef(height,weight)
```

    [
        [(height,height),(height,weight)]

```python
r=np.corrcoef(height,weight)[0,1]
r
```

```python
plt.figure()
plt.scatter(height, weight, label=f"r={r:.2f}")
plt.legend()
plt.grid()
plt.show()
```

```python
def corr_plot(X,y): 
    r=np.corrcoef(height,weight)[0,1]
    plt.figure()
    plt.scatter(X, y, label=f"r={r:.2f}")
    plt.legend()
    plt.grid()
    plt.show()
```

```python
corr_plot(height,weight)
```

```python
height=np.random.normal(150,20,100) 
weight=0.6*height + np.random.random(100)*10
```

```python
corr_plot(height, weight)
```
