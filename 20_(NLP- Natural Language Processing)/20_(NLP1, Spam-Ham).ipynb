```python
def vocab(courpus): 
    vocab=[]
    for document in courpus: 
        for word in document.split(): 
            if word not in vocab: 
                vocab.append(word)
    
    vocab.sort()
    v={w:v for v, w in enumerate(vocab)}
    return v
```

```python
vocablary=vocab(courpus)
vocablary
```

```python
courpus
```

```python
len(vocablary)
```

```python
def bag_of_word(courpus, vocablary): 
    vectors=[]
    for document in courpus: 
        vector=[0]*len(vocablary)
        for word in document.split(): 
            if word in vocablary: 
                i=vocablary[word]
                vector[i]+=1
        vectors.append(vector)
    return vectors
```

```python
bag_of_word(courpus, vocablary)
```

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
data=pd.read_csv('spam_ham.csv', encoding='latin', usecols=[0,1])
data.rename({'v1':'label', 'v2':'message'}, axis=1, inplace=True)
```

```python
data
```

```python
X=data['message']
X
```

```python
X=data['message'].str.lower()
X
```

```python
y=data['label']
y
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
vector=CountVectorizer()
vector.fit(X) 
X_tr=vector.transform(X)
```

```python
X_tr
```

```python
X_tr.toarray()
```

```python
vector.vocabulary_
```

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
model=MultinomialNB()
model.fit(X_tr,y)
y_hat=model.predict(X_tr)
```

```python
y_hat
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
print(f"Accuracy Score: {accuracy_score(y,y_hat)}")
```

```python
y
```

```python
y.value_counts()
```

```python
print(f"Confusion Matrix: \n {confusion_matrix(y, y_hat)}")
```

```python
from sklearn.model_selection import StratifiedShuffleSplit
```

```python
split=StratifiedShuffleSplit(n_splits=1, test_size=0.2, random_state=123)
```

```python
(train_idx, test_idx),=split.split(X,y)
```

```python
train_idx
```

```python
test_idx
```

```python
X_train = X[train_idx]
X_test = X[test_idx]
y_train = y[train_idx]
y_test = y[test_idx]
```

```python
from sklearn.pipeline import Pipeline
```

```python
model = Pipeline([
    ('vectorizer', CountVectorizer()),
    ('classification', MultinomialNB())
])
```

```python
model.fit(X_train, y_train)
```

```python
y_hat_train = model.predict(X_train)
y_hat_test = model.predict(X_test)
```

```python
y_hat_train
```

```python
y_hat_test
```

```python
def evaluate(y, y_hat):
    print(f"Accuracy Score: {accuracy_score(y, y_hat):.2f}\n")
    print(f"Confusion Matrix:\n {confusion_matrix(y, y_hat)}\n")
    print(f"Classification Report:\n {classification_report(y, y_hat)} ")
```

```python
evaluate(y_train, y_hat_train)
```

```python
y_train.value_counts()
```

```python
evaluate(y_test, y_hat_test)
```
