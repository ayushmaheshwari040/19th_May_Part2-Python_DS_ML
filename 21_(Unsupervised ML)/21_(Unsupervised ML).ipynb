# Unsupervised Machine Learning

- Unsupervised machine learning is a category of machine learning that focuses on discovering hidden patterns, structures, and relationships within unlabeled datasets without the need for human supervision or predefined output labels.
- Unsupervised learning algorithms are given raw, unclassified data and are tasked with autonomously identifying inherent groupings, similarities, or anomalies.
