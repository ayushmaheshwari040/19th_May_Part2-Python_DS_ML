# Some Points in Dimensionality Reduction

1. It is a process of Reducing the number of features , while keeping the useful information.
2. Benefits:-
   - Speed up the computation.
   - Reduce the storage.
   - Improve the model performance.
   - Data Visulization
3. Techniques:-
   - PCA(Principal Component Analysis) -> Convert features into new features(Principal Components).
   - UMAP(Uniform Manifold Approximation and Projection)/T.SNE -> Used for visualizing the higher dimesional data.

```python
import pandas as pd 
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
```

```python
data=load_iris()
data
```

```python
df=pd.DataFrame(data.data, columns=data.feature_names)
df
```

```python
model = PCA(n_components=2)
model
```

```python
reduced_data = model.fit_transform(df)
reduced_data                                         # principal components
```

```python
data.target
```

```python
plt.scatter(reduced_data[:, 0], reduced_data[:, 1], c=data.target)
plt.xlabel('PC1', size=15)
plt.ylabel('PC2', size=15)
plt.show()
```
