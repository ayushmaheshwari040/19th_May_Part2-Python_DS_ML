# **Machine Learning Intro**

### **Machine Learning**

- A computer program which improves its performence P on a given task T with experience E to give better outcome is known as ML. 

### **Types of ML**

1. Supervised ML
2. Unsupervised ML
3. Semisupervised ML
4. Reinforcement ML

### **Basic Terms** 

1. Population-> Complete Dataset
2. Sample-> A part of population
3. Feature-> Independent Variables
4. Target/Label-> Dependent Variables
5. Train Dataset-> A part of data by which we train our ML Model or it is learning phase.
6. Test Dataset-> A part of data by which we test our ML Model and also called Unseen Data or it is testing phase.
7. Sampling-> When we divide population in training and testing data.

### **Supervised ML**

- In Supervised ML Target/Labels are also included in Model Training.
- Ex:- Spam or Ham Classifier

#### Terms of Supervised ML 
- Regression
- Classification

### **Unsupervised ML**

- In Supervised ML Target/Labels are not included in Model Training.
- Divide the data according to their characteristics into clusters.

#### Terms of Unsupervised Learning
- Clustering
- Segmentation
- Semisupervised ML
- Anolmay Detection
- Data Leveling

### **Reinforcement Learning**

- Reward & Punishment Based
- Ex:- ChatGpt response as feedback
